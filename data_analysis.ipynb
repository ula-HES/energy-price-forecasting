# Data Analysis

This notebook contains code for data analysis - visualisations and statistical tests.

## Data import

```python
import sqlite3

# connect to database 

connection = sqlite3.connect("db_data_hoarder_ula.db")

cur = connection.cursor()


cur.execute("SELECT * FROM tge_rdnquarters_weather_forecast_kse")
rows_tge_rdnquarters_weather_forecast_kse = cur.fetchall()
cols_tge_rdnquarters_weather_forecast_kse = [desc[0] for desc in cur.description]

connection.close()
```

```python
# convert to a pandas data frame

import pandas as pd

tge_rdnquarters_weather_forecast_kse = pd.DataFrame(rows_tge_rdnquarters_weather_forecast_kse, columns = cols_tge_rdnquarters_weather_forecast_kse)
```

```python
tge_rdnquarters_weather_forecast_kse = tge_rdnquarters_weather_forecast_kse[["Price_ii_previous_day", "Price_ii_day_ahead", "Temperatura_forecast", "Wilgotność_forecast", "Opady_forecast", "Nasłonecznienie_forecast", "Prędkość_wiatru_forecast", "load_fcst" ]]
```
